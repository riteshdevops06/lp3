````python
def fracKnapsack(weights,values,capacity):
    n=len(values);
    ratio=[]
    for i in range(n):
        if(weights[i]==0):
            print(f"Skipping item {i+1} as weight is 0")
            continue
        ratio.append((values[i]/weights[i],i))
    ratio.sort(reverse=True)
    total_value=0.0
    fractions=[0]*n
    for r, i in ratio:
        if(capacity==0):
            break
        elif weights[i]<=capacity:
            total_value+=values[i]
            capacity-=weights[i]
            fractions[i]=1
        else:
            fraction=capacity/weights[i]
            total_value+=values[i]*fraction
            fractions[i]=fraction
            capacity=0
    return fractions,total_value
    
````

````python
# Sure! Let’s calculate **time and space complexity** of your `fracKnapsack` function **step by step**, using an example input to see why it is what it is.

# ---

# ## **1️⃣ Function Recap**

# ```python
# def fracKnapsack(weights, values, capacity):
#     n = len(values)
#     ratio = []
#     for i in range(n):
#         if weights[i] == 0:
#             continue
#         ratio.append((values[i]/weights[i], i))
#     ratio.sort(reverse=True)
    
#     total_value = 0.0
#     fractions = [0]*n
    
#     for r, i in ratio:
#         if capacity == 0:
#             break
#         elif weights[i] <= capacity:
#             total_value += values[i]
#             capacity -= weights[i]
#             fractions[i] = 1
#         else:
#             fraction = capacity/weights[i]
#             total_value += values[i]*fraction
#             fractions[i] = fraction
#             capacity = 0
#     return fractions, total_value
# ```

# ---

# ## **2️⃣ Step by Step Time Complexity**

# ### **Step 2.1: Compute ratios**

# ```python
# for i in range(n):
#     if weights[i] == 0: continue
#     ratio.append((values[i]/weights[i], i))
# ```

# * **Loop runs n times** → `O(n)`
# * Each iteration does **constant work** (division + append) → `O(1)`
# * So **this step = O(n)**

# ---

# ### **Step 2.2: Sort ratios**

# ```python
# ratio.sort(reverse=True)
# ```

# * Sorting **list of n elements** → `O(n log n)`

# ---

# ### **Step 2.3: Greedy selection loop**

# ```python
# for r, i in ratio:
#     ...
# ```

# * Worst case: we loop through **all n items** → `O(n)`

# * Inside loop: **constant operations** → adding, subtracting, fraction calculation → `O(1)`

# * So this loop = `O(n)`

# ---

# ### **Step 2.4: Combine all steps**

# * Ratio calculation = `O(n)`

# * Sorting = `O(n log n)`

# * Greedy loop = `O(n)`

# * **Total time complexity = O(n log n)** (sorting dominates)

# ---

# ## **3️⃣ Space Complexity**

# * `ratio` list → stores up to **n tuples** → `O(n)`
# * `fractions` list → stores **n fractions** → `O(n)`
# * Other variables → O(1)

# ✅ **Total space complexity = O(n)**

# ---

# ## **4️⃣ Dry Run With Input**

# **Input:**

# ```python
# weights = [10, 20, 30]
# values = [60, 100, 120]
# capacity = 50
# ```

# * **n = 3**

# ### **Step 1: Compute ratio**

# * ratio = `[(6.0,0), (5.0,1), (4.0,2)]` → O(n) = 3 steps

# ### **Step 2: Sort ratio**

# * Already sorted → O(n log n) = O(3 log 3) ≈ 4-5 operations

# ### **Step 3: Take items**

# * Loop through all 3 items → O(n) = 3 steps

# * So for **small n**, total operations ≈ 3 + 5 + 3 = 11 operations

# * For large n, it scales as **O(n log n)**

# * Space: ratio list = 3 elements, fractions list = 3 elements → O(n) = 6 elements

# ---

# ### ✅ **Summary**

# | Step                  | Complexity | Why                         |
# | --------------------- | ---------- | --------------------------- |
# | Ratio calculation     | O(n)       | Loop through n items        |
# | Sorting ratios        | O(n log n) | Sorting dominates           |
# | Greedy selection loop | O(n)       | Loop through sorted items   |
# | **Total time**        | O(n log n) | Sorting dominates           |
# | Space used            | O(n)       | ratio list + fractions list |

# ---

# If you want, I can **also show a “step by step counter” version**, counting **operations for each line** using this example, so you can see **exactly how many steps are executed**.

# Do you want me to do that?


# Got it! Let’s break this **carefully** so you understand **why greedy works for Fractional Knapsack**, what other techniques exist, and why **other methods are not necessary here**.

# ---

# ## **1️⃣ Why Greedy is used here**

# * **Problem:** Fractional Knapsack → you **can take fractions of items**.
# * **Goal:** Maximize total value without exceeding capacity.

# **Greedy strategy:**

# 1. Compute **value/weight ratio** for each item → “profit per unit weight”
# 2. Pick items **starting with highest ratio**
# 3. If an item doesn’t fit fully → take the **fraction that fits**

# **Why greedy works:**

# * For fractional items, taking the item with **highest value per weight first** **always maximizes profit**.
# * This is provably **optimal** using the **greedy-choice property**:

# > At each step, the **locally best choice (highest ratio)** leads to the **global optimum**.

# * This is **not true for 0/1 Knapsack** (cannot take fractions), but **fractional knapsack allows it**, so greedy is enough.

# ---

# ## **2️⃣ Could other techniques be used?**

# Technically, yes, but **they are not needed** because greedy is optimal:

# 1. **Dynamic Programming (DP)**

#    * Works for **0/1 Knapsack**
#    * Can be used for fractional knapsack, but **overkill**
#    * DP would be **slower and more complex** (O(n*capacity)), greedy is faster (O(n log n))

# 2. **Brute Force / Recursion**

#    * Try all possible fractions → impractical because **fractions are infinite possibilities**
#    * Greedy avoids this by taking **maximum possible fraction at each step**

# 3. **Branch and Bound**

#    * Used for 0/1 Knapsack for pruning non-promising branches
#    * Fractional knapsack doesn’t need pruning → greedy gives exact solution immediately

# ---

# ## **3️⃣ Why greedy only for Fractional Knapsack**

# * **Greedy works only if items can be split.**
# * Proof idea: If you take the item with **highest value/weight ratio first**, you can’t get more profit by choosing any other item instead.
# * Complexity is low:

#   * **O(n)** for ratio calculation
#   * **O(n log n)** for sorting → total = O(n log n)
# * No DP or recursion is needed → simpler and faster

# ---

# ### **4️⃣ Comparison**

# | Method         | Works for Fractional? | Optimal? | Complexity                |
# | -------------- | --------------------- | -------- | ------------------------- |
# | Greedy (ratio) | ✅                     | ✅        | O(n log n)                |
# | DP Table       | ✅                     | ✅        | O(n*capacity) → slower    |
# | Brute Force    | ✅                     | ✅        | Exponential → impractical |
# | Branch & Bound | ✅                     | ✅        | More complex              |

# ---

# ### ✅ **Key takeaway**

# * **Fractional Knapsack → greedy is optimal** because of **value/weight ratio**
# * Other techniques exist, but **greedy is simpler, faster, and exact**
# * **0/1 Knapsack → greedy doesn’t work**, DP or branch & bound needed

# ---

# If you want, I can **illustrate with an example showing what happens if we don’t use greedy**, to prove why greedy is the best choice for fractional knapsack.

# Do you want me to do that?
````
