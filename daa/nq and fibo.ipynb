````python
def print_solution(board,n):
    for i in range(n):
        for j in range(n):
            print(board[i][j],end=" ")
        print()
    print()
````

````python
def isSafe(board, row, col,n):
    for i in range(row):
        if board[i][col]==1:
            return False

    i,j=row-1,col-1
    while i>=0 and j>=0:
        if board[i][j]==1:
            return False
        i-=1
        j-=1

    i,j=row-1,col+1
    while i>=0 and j<n:
        if board[i][j]==1:
            return False
        i-=1
        j+=1
    return True
````

````python
def solve(board,row,n):
    if row==n:
        print_solution(board,n)
        return
        
    for col in range(n):
        if isSafe(board,row,col,n):
            board[row][col]=1
            solve(board,row+1,n)
            board[row][col]=0
        
````

````python
def n_queens(n,first_row,first_column):
    board=[[0 for _ in range(n)] for _ in range(n)]

    board[first_row][first_column]=1
    solve(board,first_row+1,n)
````

````python
n=int(input("Enter the value of n"))
r=int(input("Enter the row number for the 1 st queen (0 to n-1)"))
c=int(input("Enter the column number for the 1 st queen (0 to n-1)"))
n_queens(n,r,c)
````

````python
# Ah! Let’s carefully explain **why the optimized N-Queens solution has O(n!) time complexity** instead of O(n^n). We’ll go **step by step**.

# ---

# ## **1️⃣ Naive recursion (O(n^n))**

# * Each row tries **all n columns** → n choices per row
# * Recursion depth = n
# * Total combinations = **n × n × … × n (n times) = n^n**

# ✅ That’s why the naive upper bound is O(n^n).

# ---

# ## **2️⃣ Why O(n!) in optimized version**

# * **Observation:** Each column can have **at most one queen** (because two queens in same column attack each other).

# * So after placing a queen in **row 0**, that column is **blocked** for all remaining rows.

# * Row 0 → n choices

# * Row 1 → n-1 choices (one column is blocked)

# * Row 2 → n-2 choices

# * …

# * Row n-1 → 1 choice

# * Total number of placements = **n × (n-1) × (n-2) × … × 1 = n!**

# ---

# ### **3️⃣ Using diagonals (optimization)**

# * With extra arrays to track diagonals:

# ```python
# cols = [False]*n
# diag1 = [False]*(2*n-1)  # row+col
# diag2 = [False]*(2*n-1)  # row-col+n-1
# ```

# * `isSafe()` becomes O(1) instead of O(n)
# * Time per recursive call = O(1)
# * Number of calls = O(n!) → **total time = O(n!)**

# ---

# ### **4️⃣ Intuition**

# * **Columns constraint** reduces choices drastically.

# * Without this, each row has **n options** → O(n^n)

# * With column tracking → first row n choices, second row n-1 choices, … last row 1 choice → **O(n!)**

# * Diagonal constraints further **prune impossible branches**, but **O(n!) is the typical worst-case bound**.

# ---

# ### ✅ **Summary**

# | Approach                                       | Time Complexity | Why                                                          |
# | ---------------------------------------------- | --------------- | ------------------------------------------------------------ |
# | Naive recursion (check every column)           | O(n^n)          | Each row has n choices                                       |
# | Optimized recursion + column/diagonal tracking | O(n!)           | Each row has fewer available columns → multiply choices → n! |

# ---

# If you want, I can **draw the n=4 recursion tree with pruning** to **show exactly how choices reduce from 4 → 3 → 2 → 1**, which makes O(n!) very easy to visualize.

# Do you want me to do that?
````

````python

 
## ---------- Non Recursive ------------ ##
 
def fib_iterative(n):
    if n <= 1:
        return n
    a, b = 0, 1
    for i in range(2, n + 1):
        c = a + b
        a = b
        b = c
    return b
 
 
n = int(input("Enter n: "))
print(f"Fibonacci({n}) = {fib_iterative(n)}")
 
 
## ---------- Recursive ------------ ##
 
def fib_recursive(n):
    if n <= 1:
        return n
    return fib_recursive(n - 1) + fib_recursive(n - 2)
 
n = int(input("Enter n: "))
print(f"Fibonacci({n}) = {fib_recursive(n)}")
 
 
## ---------- TOH ------------ ##
 
def tower_of_hanoi(n, source, target,auxiliary):
    if n == 1:
        print(f"Move disk 1 from {source} to {target}")
        return
    # Step 1: Move n-1 disks from source → auxiliary
    tower_of_hanoi(n - 1, source, auxiliary, target)
    # Step 2: Move remaining disk from source → target
    print(f"Move disk {n} from {source} to {target}")
    # Step 3: Move n-1 disks from auxiliary → target
    tower_of_hanoi(n - 1, auxiliary, source, target)
 
n = int(input("Enter number of disks: "))
tower_of_hanoi(n, 'A', 'B', 'C')
````
