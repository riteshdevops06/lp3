````python
import heapq
from collections import defaultdict
````

````python
class Node:
    def __init__(self,char,freq):
        self.char=char
        self.freq=freq
        self.left=None
        self.right=None
        
    def __lt__(self,other):
        return self.freq<other.freq
````

````python
def build_huffmanTree(text):
    freq=defaultdict(int)
    for ch in text:
        freq[ch]+=1
        
    heap=[Node(ch,f) for ch,f in freq.items()]
    heapq.heapify(heap)

    while len(heap)>1:
        left=heapq.heappop(heap)
        right=heapq.heappop(heap)
        merged=Node(None,left.freq+right.freq)
        merged.left=left
        merged.right=right
        heapq.heappush(heap,merged)
    return heap[0]
````

````python
def generate_codes(root,current_code="",codes={}):
    if root is None:
        return
    if root.char is not None:
        codes[root.char]=current_code
        return
        
    generate_codes(root.left,current_code+"0",codes)
    generate_codes(root.right,current_code+"1",codes)
    return codes
````

````python
def encode(text,codes):
    return ''.join(codes[ch] for ch in text) 
````

````python
def decode(encoded_text, root):
    decoded = ""
    node = root
    for bit in encoded_text:
        node = node.left if bit == "0" else node.right
        if node.char:
            decoded += node.char
            node = root
    return decoded
````

````python
if __name__=="__main__":
    text=input("Enter the text : ")
    root=build_huffmanTree(text)
    codes=generate_codes(root)
    for ch, code in codes.items():
        print(f"{repr(ch)}:{code}")
    encoding=encode(text,codes)
    print(f"The encoded text is : {encoding}")
    decoding=decode(encoding,root)
    print(f"The decoded text is : {decoding}")
````

````python
# Excellent — that’s exactly the kind of deep question that shows real understanding 👏

# Let’s go step-by-step and explain **why Huffman coding uses a *greedy algorithm*** and **not backtracking or other methods**, and then break down what “greedy” really means.

# ---

# ## 🧠 Step 1: What Is a *Greedy Algorithm*?

# A **Greedy Algorithm** means:

# > At each step, we make the **best immediate choice** (locally optimal), hoping it leads to a **global optimum** (overall best solution).

# ✅ **In short:**
# Take the best step *right now*, don’t look back, and don’t explore all possibilities.

# ---

# ### ⚙️ Example:

# Imagine you want to give change of ₹36 using ₹20, ₹10, ₹5, and ₹1 coins.
# A greedy algorithm will pick:

# * One ₹20 → remaining ₹16
# * One ₹10 → remaining ₹6
# * One ₹5 → remaining ₹1
# * One ₹1 → done ✅

# → Total coins = 4
# This is the **optimal** solution, found greedily (without exploring every combination).

# ---

# ## 🧩 Step 2: How Huffman Coding Is *Greedy*

# In Huffman coding, at every step you:

# 1. **Choose the two smallest frequency nodes**,
# 2. **Merge them** into one node,
# 3. **Repeat** until only one node (the root) remains.

# ### 🔁 Each step makes a *locally optimal* choice:

# > Combine the two least frequent symbols first.

# Because less frequent symbols should have **longer codes**, and merging them first ensures they end up **deeper in the tree** (hence longer binary codes).

# ---

# ### 📈 Why This Works (Proof Sketch)

# The *greedy choice property* holds because:

# * In any optimal prefix code, the **two least frequent symbols** will always be **siblings at the deepest level** of the tree.
# * Huffman’s algorithm enforces that property at each step.

# So making the *best local choice* (merge smallest two) leads to the *globally optimal* coding tree.
# That’s exactly what defines a **greedy algorithm**.

# ---

# ## 🚫 Step 3: Why Not *Backtracking* or *Dynamic Programming*?

# Let’s analyze alternatives:

# | Approach                | Description                                              | Why Not Used Here                                                                         |
# | ----------------------- | -------------------------------------------------------- | ----------------------------------------------------------------------------------------- |
# | **Backtracking**        | Try all possible combinations, backtrack if not optimal. | Too slow — exponential time (`O(2^k)`), since it would test all possible tree structures. |
# | **Dynamic Programming** | Solve subproblems and combine optimal sub-results.       | There’s no overlapping subproblem structure — each merge step changes the tree uniquely.  |
# | **Greedy** ✅            | Choose smallest two frequencies repeatedly.              | Fast (`O(k log k)`) and *guaranteed optimal* (proved by Huffman).                         |

# So Huffman coding fits **Greedy**, not Backtracking or DP, because:

# * It builds solution step-by-step without revisiting prior decisions.
# * The problem satisfies both:

#   * **Greedy-choice property**, and
#   * **Optimal substructure** (the merged tree is also optimal).

# ---

# ## 🧩 Step 4: Summary of Why Huffman = Greedy

# | Feature                  | Description                                          |
# | ------------------------ | ---------------------------------------------------- |
# | **Choice at each step**  | Merge two smallest frequencies                       |
# | **No backtracking**      | Never reconsiders earlier merges                     |
# | **Optimal substructure** | Subtrees built optimally remain optimal in full tree |
# | **Result**               | Minimum total weighted path length                   |
# | **Complexity**           | `O(k log k)` (using min-heap)                        |

# ---

# ### ✅ **In Simple Words:**

# > Huffman coding is *greedy* because it keeps making the smallest possible merge at each step, ensuring frequent symbols get shorter codes and rare symbols get longer ones — achieving optimal compression **without exploring all possibilities**.

# ---

# Would you like me to show a **small visual Huffman tree** example to illustrate exactly how the greedy step (picking two smallest) affects the tree’s structure and encoding length?
````
