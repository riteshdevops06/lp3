````python
class Item:
    def __init__(self,value,weight):
        self.value=value
        self.weight=weight
````

````python
def print_dp(dp):
    for row in dp:
        for val in row:
            print(f"{val:4}",end=" ")
        print()
    print()
````

````python
def knapsack(capacity,items):
    n=len(items)
    dp=[[0]*(capacity+1) for _ in range(n+1)]

    for i in range(1,n+1):
        for w in range(1,capacity+1):
            if items[i-1].weight<=w:
                dp[i][w]=max(dp[i-1][w],items[i-1].value+dp[i-1][w-items[i-1].weight])
            else:
                dp[i][w]=dp[i-1][w]
    print_dp(dp)

    result=dp[n][capacity]
    w=capacity

    for i in range(n,0,-1):
        if result==dp[i-1][w]:
            continue
        else:
            print(f"Item(weight={items[i - 1].weight}, value={items[i - 1].value})")
            result-=items[i-1].value
            w-=items[i-1].weight
    return dp[n][capacity]
````

````python
n = int(input("Enter the number of items: "))
items = []
print("Enter value and weight of each item:")
for i in range(n):
    val = int(input(f"Enter the value of item {i+1}: "))
    wei = int(input(f"Enter the weight of item {i+1}: "))  # Corrected line
    items.append(Item(val, wei))

capacity = int(input("Enter the capacity of the knapsack: "))
max_profit = knapsack(capacity, items)
print(f"Maximum profit obtained is {max_profit}")
````

````python
# Perfect question, Rohan — this is the **classic 0/1 Knapsack Problem**, and your implementation is the **Dynamic Programming (DP)** solution.

# Let’s analyze it step by step 👇

# ---

# ## 🧠 Step 1: Understanding the Problem

# You have:

# * `n` = number of items
# * `capacity` = maximum weight the knapsack can carry

# Each item has:

# * `weight`
# * `value`

# Goal:

# > Choose a subset of items to **maximize total value** without exceeding the capacity.

# ---

# ## ⚙️ Step 2: How the Code Works

# You use a **2D DP table**:

# ```python
# dp = [[0] * (capacity + 1) for _ in range(n + 1)]
# ```

# Where:

# * `dp[i][w]` = maximum value we can get using the **first i items** with **capacity w**

# Then, you fill it with:

# ```python
# for i in range(1, n + 1):
#     for w in range(1, capacity + 1):
#         if items[i-1].weight <= w:
#             dp[i][w] = max(dp[i-1][w],
#                            items[i-1].value + dp[i-1][w - items[i-1].weight])
#         else:
#             dp[i][w] = dp[i-1][w]
# ```

# So:

# * You **either take the item** or **skip it**, whichever gives more value.
# * This builds up all possible solutions in a **bottom-up** way.

# ---

# ## ⏱ Step 3: **Time Complexity**

# Let’s break it down:

# * Two nested loops:

#   ```python
#   for i in range(1, n + 1):         # runs n times
#       for w in range(1, capacity + 1):  # runs capacity times
#   ```

#   ⇒ Total iterations = `n * capacity`

# * Each iteration does `O(1)` work (a simple comparison and assignment).

# ✅ **Total Time Complexity:**
# [
# O(n \times capacity)
# ]

# ---

# ## 💾 Step 4: **Space Complexity**

# You created a 2D table:

# ```python
# dp[n+1][capacity+1]
# ```

# → Total cells = `(n + 1) * (capacity + 1)`

# ✅ **Space Complexity:**
# [
# O(n \times capacity)
# ]

# ---

# ## 🧩 Step 5: Why Use **Dynamic Programming (DP)** Instead of Other Techniques

# Let’s compare approaches 👇

# | Approach                    | Idea                                                                  | Time                            | Why Not Used                                              |
# | --------------------------- | --------------------------------------------------------------------- | ------------------------------- | --------------------------------------------------------- |
# | **Brute Force / Recursion** | Try all combinations (take / skip each item).                         | `O(2^n)`                        | Exponential — too slow even for moderate n (~30).         |
# | **Backtracking**            | Same as recursion but with pruning (skip paths if weight > capacity). | Still exponential in worst case | Might reduce work, but not guaranteed optimal efficiency. |
# | **Greedy**                  | Pick items based on value/weight ratio (like fractional knapsack).    | `O(n log n)`                    | Doesn’t work for 0/1 case — local best ≠ global best.     |
# | **Dynamic Programming** ✅   | Build optimal solution using previous subproblems.                    | `O(n × capacity)`               | Efficient and always optimal.                             |

# ---

# ## 🎯 Step 6: Why DP Works (Key Properties)

# DP applies here because the problem has:

# ### 1️⃣ **Optimal Substructure**

# > The optimal solution for `n` items and capacity `W` depends on:
# >
# > * Optimal solution for `n-1` items and capacity `W`
# > * Optimal solution for `n-1` items and capacity `W - weight[n-1]`

# Mathematically:
# [
# dp[i][w] = \max(dp[i-1][w],\ value[i-1] + dp[i-1][w-weight[i-1]])
# ]

# ---

# ### 2️⃣ **Overlapping Subproblems**

# > Many subproblems repeat (e.g., computing `dp[i-1][w]` for many i).

# DP **stores** these sub-results instead of recomputing → huge efficiency gain over recursion.

# ---

# ## 💡 Step 7: Optimization Tip

# You can reduce space from `O(n × capacity)` → `O(capacity)` by using a **1D DP array**:

# ```python
# dp = [0] * (capacity + 1)
# for item in items:
#     for w in range(capacity, item.weight - 1, -1):
#         dp[w] = max(dp[w], item.value + dp[w - item.weight])
# ```

# ✅ **Time:** still `O(n × capacity)`
# ✅ **Space:** now `O(capacity)`

# ---

# ## ✅ Final Summary

# | Metric          | Explanation                                                  | Complexity                    |
# | --------------- | ------------------------------------------------------------ | ----------------------------- |
# | **Time**        | Two nested loops                                             | **O(n × capacity)**           |
# | **Space**       | 2D DP table                                                  | **O(n × capacity)**           |
# | **Why DP**      | Has optimal substructure + overlapping subproblems           | Most efficient exact solution |
# | **Alternative** | Greedy (fails), Backtracking (slow), Recursion (exponential) | ❌                             |

# ---

# Would you like me to show how the **DP table builds up step-by-step visually** for a small example (say 3 items, capacity = 5)? It’s a great way to *see* why DP works.
````

````python
# Absolutely! Let’s go **carefully** and compare **Dynamic Programming (DP)**, **Greedy**, and **Backtracking**, explaining what they are, when they are used, and how they work. I’ll also show **examples** so it’s very clear.

# ---

# # **1️⃣ Dynamic Programming (DP)**

# ### **Idea:**

# * Solve **complex problems by breaking them into smaller subproblems**.
# * Store the results of subproblems (**memoization or table**) to **avoid recomputation**.
# * Works best for problems with **optimal substructure** and **overlapping subproblems**.

# ---

# ### **When to use:**

# * Example: **0/1 Knapsack** (cannot take fractions)
# * Subproblems: “Maximum value using first i items with capacity w”

# ---

# ### **How it works (0/1 Knapsack example):**

# 1. Create a DP table `dp[n+1][capacity+1]`

#    * Rows = items considered
#    * Columns = capacity
# 2. Fill table row by row:

# ```python
# if item.weight <= w:
#     dp[i][w] = max(dp[i-1][w], dp[i-1][w-item.weight] + item.value)
# else:
#     dp[i][w] = dp[i-1][w]
# ```

# 3. `dp[n][capacity]` → maximum profit
# 4. Trace back the table to find which items were included

# ---

# ### **Pros:**

# * Exact solution
# * Works for 0/1 Knapsack and many combinatorial problems

# ### **Cons:**

# * Can use **a lot of memory** (O(n*capacity))
# * Slower than greedy for problems where greedy is optimal

# ---

# # **2️⃣ Greedy Algorithm**

# ### **Idea:**

# * Always make the **locally best choice** at each step
# * Hope that local choices lead to **global optimum**

# ---

# ### **When to use:**

# * Works **perfectly for Fractional Knapsack** (items can be divided)
# * Works for **Huffman coding, activity selection, coin change (certain currencies)**

# ---

# ### **How it works (Fractional Knapsack example):**

# 1. Compute **value/weight ratio** for all items
# 2. Sort items **descending by ratio**
# 3. Take items in order:

#    * If it fits → take full
#    * If it doesn’t → take fraction
# 4. Stop when capacity = 0

# ---

# ### **Pros:**

# * Very fast → O(n log n) because of sorting
# * Simple to implement
# * Optimal for fractional knapsack

# ### **Cons:**

# * **Not guaranteed optimal for 0/1 Knapsack**
# * Only works when **greedy-choice property** holds

# ---

# # **3️⃣ Backtracking**

# ### **Idea:**

# * Explore **all possibilities recursively**
# * “Backtrack” when a partial solution **cannot lead to a better result**

# ---

# ### **When to use:**

# * Exact solution for **0/1 Knapsack**
# * Solving **permutation, combination, subset sum, n-queens**

# ---

# ### **How it works (0/1 Knapsack example):**

# 1. Start from first item
# 2. For each item, **branch into two choices**:

#    * Include the item → reduce remaining capacity
#    * Exclude the item → keep capacity same
# 3. Recursively compute max value
# 4. Optionally use **Branch and Bound** to **prune branches** that cannot beat current best

# ---

# ### **Pros:**

# * Gives exact solution
# * Can handle problems where DP is hard to define

# ### **Cons:**

# * Very slow → O(2^n) for n items
# * Needs pruning to be practical

# ---

# # **4️⃣ Comparison Table**

# | Method          | Works for 0/1 Knapsack | Works for Fractional Knapsack | Time Complexity | Space Complexity | Notes                                                 |
# | --------------- | ---------------------- | ----------------------------- | --------------- | ---------------- | ----------------------------------------------------- |
# | DP (Tabulation) | ✅                      | ❌ (can modify but slower)     | O(n*capacity)   | O(n*capacity)    | Optimal, but uses memory                              |
# | Greedy          | ❌                      | ✅                             | O(n log n)      | O(n)             | Fast and simple, optimal only if fractional items     |
# | Backtracking    | ✅                      | ✅                             | O(2^n)          | O(n) recursion   | Exact, but exponential unless pruned (branch & bound) |

# ---

# # **5️⃣ Quick Examples**

# **Fractional Knapsack (Greedy)**

# | Item | Value | Weight | Ratio |
# | ---- | ----- | ------ | ----- |
# | 1    | 60    | 10     | 6.0   |
# | 2    | 100   | 20     | 5.0   |
# | 3    | 120   | 30     | 4.0   |

# * Capacity = 50 → take item 1 (10), item 2 (20), 2/3 of item 3 → total value = 240

# ---

# **0/1 Knapsack (DP)**

# * Items cannot be divided → pick items **1 and 3** or **2 and 3** depending on DP table → find max value using table
# * DP table tracks **which items fit** for each capacity

# ---

# **0/1 Knapsack (Backtracking)**

# * Explore all **8 possibilities** (2^3 items) → pick combination giving **max value ≤ capacity**
# * Can prune branches with **branch & bound**

# ---

# ### ✅ **Key Takeaways**

# 1. **Greedy** → for **fractional knapsack**, fast and optimal
# 2. **DP** → for **0/1 knapsack**, exact solution, uses table
# 3. **Backtracking** → exact solution, explores all possibilities, can prune with branch & bound

# ---

# If you want, I can **draw a single visual diagram** showing **DP table, greedy selection, and backtracking tree side by side** for the same example. It helps to **see all three methods at once**.

# Do you want me to do that?
````
