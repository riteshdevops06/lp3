```python
def function(x):
    return (x+3)**2
```

```python
def derivative(x):
    return 2*(x+2)
```

```python
max_iterations=1000
precision=0.0001
learning_rate=0.1
x=2
```

```python
for i in range(max_iterations):
    grad=derivative(x)
    x_new=x-learning_rate*grad
    if(abs(x_new-x)<precision):
        print(f"Converged at iteration: {i}")
        break
    x=x_new
```

```python
print(f"Local minima occurs at x={x}")
print(f"Value of the function at local minimum = {function(x)}")
```
