```python
df.drop('Email No.',axis=1,inplace=True)
```

```python
df.head()
```

```python
x=df.drop('Prediction',axis=1)
y=df['Prediction']
```

```python
x.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler,Normalizer
```

```python
sc=StandardScaler()
x_train_svm=sc.fit_transform(x_train)
x_test_svm=sc.transform(x_test)
```

```python
norm=Normalizer()
x_train_knn=norm.fit_transform(x_train)
x_test_knn=norm.transform(x_test)
```

```python
from sklearn.svm import SVC
```

```python
svc=SVC(kernel='linear',class_weight='balanced',random_state=42)
svc.fit(x_train_svm,y_train)
svcPred=svc.predict(x_test_svm)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn=KNeighborsClassifier(n_neighbors=5,metric='euclidean')
knn.fit(x_train_knn,y_train)
knnPred=knn.predict(x_test_knn)
```

```python
from sklearn.metrics import classification_report,accuracy_score
```

```python
print(f"SVM accuracy is {accuracy_score(y_test,svcPred)}")
print(f"SVM report is {classification_report(y_test,svcPred)}")
```

```python
print(f"KNN accuracy is {accuracy_score(y_test,knnPred)}")
print(f"KNN report is {classification_report(y_test,knnPred)}")
```
